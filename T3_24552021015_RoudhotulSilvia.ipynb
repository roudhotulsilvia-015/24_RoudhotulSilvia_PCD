```python
## Jawaban Soal No 1
import cv2
import matplotlib.pyplot as plt

# Membaca gambar menggunakan OpenCV
gambar_bgr = cv2.imread('orange.jpg')

# Ubah ruang warna dari BGR ke RGB
gambar_rgb = cv2.cvtColor(gambar_bgr, cv2.COLOR_BGR2RGB)

# Tampilkan menggunakan Matplotlib dengan ukuran plot 6 x 6
plt.figure(figsize=(6, 6))
plt.imshow(gambar_rgb)
plt.axis('off')
plt.title('Gambar Orange (RGB)')
plt.show()
```
