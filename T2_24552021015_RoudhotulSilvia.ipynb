### **Tugas 3 — Operasi Matematika**

Buat program kalkulator sederhana dengan ketentuan:

- Minta pengguna memasukkan dua angka.
- Konversikan input menjadi `int` atau `float`.
- Hitung penjumlahan, pengurangan, perkalian, pembagian, dan pangkat.
- Tampilkan hasil setiap operasi.
- Tambahkan perhitungan luas persegi panjang berdasarkan input panjang dan lebar.

Output yang diharapkan: hasil kalkulasi berdasarkan angka yang dimasukkan pengguna.



````python
#Tugas 3

print("=" * 40)
print("     PROGRAM KALKULATOR SEDERHANA     ")
print("=" * 40)
angka1 = float(input("Masukkan angka pertama: "))
angka2 = float(input("Masukkan angka kedua : "))
print("-" * 40)
penjumlahan = angka1 + angka2
pengurangan = angka1 - angka2
perkalian = angka1 * angka2
if angka2 != 0:
    pembagian = angka1 / angka2
else:
    pembagian = "Tidak terdefinisi (pembagian dengan nol)"
pangkat = angka1 ** angka2
print("HASIL OPERASI MATEMATIKA:")
print(f"1. Penjumlahan ({angka1} + {angka2}) = {penjumlahan}")
print(f"2. Pengurangan ({angka1} - {angka2}) = {pengurangan}")
print(f"3. Perkalian   ({angka1} * {angka2}) = {perkalian}")
print(f"4. Pembagian   ({angka1} / {angka2}) = {pembagian}")
print(f"5. Pangkat     ({angka1} ^ {angka2}) = {pangkat}")
print("-" * 40)
print("PERHITUNGAN LUAS PERSEGI PANJANG:")
panjang = float(input("Masukkan panjang : "))
lebar = float(input("Masukkan lebar   : "))
luas_persegi_panjang = panjang * lebar
print(f"Luas Persegi Panjang ({panjang} x {lebar}) = {luas_persegi_panjang}")
print("=" * 40)
````

### **Tugas 4 — Perulangan `for` dan `while`**

Buat program untuk menampilkan data mahasiswa dengan ketentuan:

- Buat daftar minimal lima nama mahasiswa.
- Tampilkan semua nama menggunakan `for`.
- Tampilkan angka 1 sampai 10 menggunakan `range()`.
- Tampilkan kembali daftar nama menggunakan `while` dan `len()`.
- Buat perulangan `while` untuk menampilkan bilangan genap dari 2 sampai 20.

Output yang diharapkan: daftar nama dan deret angka yang tampil sesuai ketentuan.


````python
#Tugas 4

print("=" * 45)
print("     PROGRAM PERULANGAN FOR DAN WHILE     ")
print("=" * 45)
daftar_mahasiswa = [
    "Roudhotul Silvia",
    "Ahmad Fauzi",
    "Siti Aminah",
    "Budi Santoso",
    "Dewi Lestari"
]

print("Daftar Nama Mahasiswa (Menggunakan for):")
for i, nama in enumerate(daftar_mahasiswa, start=1):
    print(f"{i}. {nama}")
print("-" * 45)
print("Deret Angka 1 sampai 10 (Menggunakan range):")
for angka in range(1, 11):
    print(angka, end=" ")
print() # Baris baru
print("-" * 45)
print("Daftar Nama Mahasiswa (Menggunakan while & len):")
indeks = 0
while indeks < len(daftar_mahasiswa):
    print(f"- {daftar_mahasiswa[indeks]}")
    indeks += 1
print("-" * 45)
print("Bilangan Genap 2 sampai 20 (Menggunakan while):")
bilangan_genap = 2
while bilangan_genap <= 20:
    print(bilangan_genap, end=" ")
    bilangan_genap += 2
print("\n" + "=" * 45)
````

**Tugas 5 — Fungsi dan Waktu**

Buat program pencatatan aktivitas belajar dengan ketentuan:

- Impor modul `datetime`.
- Buat fungsi bernama `catat_waktu()`.
- Tampilkan pesan `Aktivitas selesai`.
- Tampilkan waktu saat fungsi dijalankan menggunakan `datetime.datetime.now()`.
- Panggil fungsi sebelum dan sesudah perulangan.
- Gunakan perulangan `for` untuk menampilkan proses pengerjaan lima soal.

Output yang diharapkan: pesan aktivitas, waktu pelaksanaan, dan proses pengerjaan soal.




````python
# Tugas 5

import datetime
def catat_waktu():
    print("Aktivitas selesai[cite: 6]")
    waktu_sekarang = datetime.datetime.now()
    format_waktu = waktu_sekarang.strftime("%d-%m-%Y %H:%M:%S")
    print(f"Waktu eksekusi: {format_waktu}\n")
print("=" * 50)
print("     PROGRAM PENCATATAN AKTIVITAS BELAJAR     ")
print("=" * 50)
print("--- Status Sebelum Memulai Pengerjaan ---")
catat_waktu()
print("--- Proses Pengerjaan Soal ---")
for nomor_soal in range(1, 6):
    print(f"Sedang mengerjakan Soal ke-{nomor_soal}...")
print("-" * 50)
print("--- Status Setelah Selesai Pengerjaan ---")
catat_waktu()
print("=" * 50)
````

## D. Tugas proyek akhir — Sistem Rekap Nilai Mahasiswa

Sebuah program digunakan untuk mencatat data mahasiswa, menerima nilai, menghitung rata-rata, dan menampilkan waktu penyelesaian proses.

Spesifikasi program:

1. Tampilkan judul program menggunakan `print()` dan garis pemisah.
2. Minta input nama dan NIM mahasiswa.
3. Minta pengguna memasukkan nilai tiga mata kuliah.
4. Hitung total dan rata-rata nilai menggunakan operasi matematika.
5. Tampilkan nama mahasiswa dalam format huruf kapital menggunakan `upper()`.
6. Gunakan `for` untuk menampilkan seluruh nilai.
7. Gunakan `while` untuk menampilkan nomor urut mata kuliah.
8. Buat fungsi `tampilkan_hasil()` untuk menampilkan total dan rata-rata.
9. Buat fungsi `catat_waktu()` untuk menampilkan waktu saat program menyelesaikan proses.
10. Tampilkan hasil akhir secara rapi dan mudah dibaca.

### Contoh gambaran output proyek akhir

```
================================
       REKAP NILAI MAHASISWA
================================
Nama       : ANDI PRATAMA
NIM        : 123456
--------------------------------
Nilai Algoritma   : 80
Nilai Basis Data  : 85
Nilai Struktur Data: 90
--------------------------------
Total Nilai : 255
Rata-rata   : 85.0
--------------------------------
Status      : Rekap selesai
Waktu       : [waktu program dijalankan]
================================
```


Data di atas hanya contoh tampilan. Mahasiswa harus membuat program yang menghasilkan nilai berdasarkan input pengguna.


````python
#Tugas D

import datetime
def tampilkan_hasil(total, rata_rata):
    print("-" * 40)
    print(f"Total Nilai : {total}")
    print(f"Rata-rata   : {rata_rata}")
def catat_waktu():
    print("-" * 40)
    print("Status      : Rekap selesai")
    waktu_sekarang = datetime.datetime.now().strftime("%d-%m-%Y %H:%M:%S")
    print(f"Waktu       : {waktu_sekarang}")
def main():
    print("=" * 40)
    print("        REKAP NILAI MAHASISWA        ")
    print("=" * 40)
    nama_input = input("Masukkan Nama Mahasiswa : ")
    nim_input = input("Masukkan NIM Mahasiswa  : ")
    nama_kapital = nama_input.upper()
    print("-" * 40)
    nilai_algoritma = float(input("Masukkan Nilai Algoritma      : "))
    nilai_basis_data = float(input("Masukkan Nilai Basis Data     : "))
    nilai_struktur_data = float(input("Masukkan Nilai Struktur Data  : "))
    daftar_nilai = [nilai_algoritma, nilai_basis_data, nilai_struktur_data]
    total_nilai = sum(daftar_nilai)
    rata_rata_nilai = total_nilai / len(daftar_nilai)
    print("-" * 40)
    print(f"Nama        : {nama_kapital}")
    print(f"NIM         : {nim_input}")
    print("-" * 40)
    mata_kuliah = ["Algoritma", "Basis Data", "Struktur Data"]
    i = 0
    while i < len(mata_kuliah):
        print(f"Nilai {mata_kuliah[i]:<14} : {daftar_nilai[i]}")
        i += 1
    tampilkan_hasil(total_nilai, rata_rata_nilai)
    catat_waktu()
    print("=" * 40)
if __name__ == "__main__":
    main()
````
