## Tugas Praktikum 1: Eksperimen Nilai Ambang Batas (Thresholding)

**Tujuan:**
Memahami pengaruh perubahan nilai ambang batas (*threshold*) terhadap hasil citra biner.

**Instruksi Tugas:**
1. Gunakan citra ***selain `kucing.jpg`***.
2. Lakukan konversi citra tersebut menjadi Grayscale.
3. Buatlah **tiga variasi** nilai ambang batas (*threshold*), yaitu:
   - $T_1 = 80$
   - $T_2 = 128$ (standar)
   - $T_3 = 200$
4. Lakukan proses thresholding untuk masing-masing nilai ambang batas tersebut.
5. Tampilkan hasil ketiga citra biner tersebut secara berdampingan dalam satu baris menggunakan `matplotlib` (gunakan fungsi `plt.subplot`).
6. Tulis analisis singkat di bawah hasil visualisasi mengenai perbedaan yang terlihat pada ketiga gambar hasil thresholding tersebut.

**CONTOH HASIL**
image.png

```python
# @title Kode untuk Eksperimen Nilai Ambang Batas (Thresholding)
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Membaca citra asli dan mengonversinya menjadi Grayscale
img = Image.open('Matahari.jpg')
gray_img = img.convert('L')
gray_array = np.array(gray_img)

# 2. Buat tiga variasi nilai ambang batas (threshold): T1 = 80, T2 = 128, T3 = 200[cite: 9]
T1 = 80
T2 = 128
T3 = 200

# Lakukan proses thresholding secara manual menggunakan NumPy array (operasi perbandingan)
# Jika nilai piksel > T, ubah jadi 255 (putih), jika tidak ubah jadi 0 (hitam)
thresh1 = np.where(gray_array > T1, 255, 0).astype(np.uint8)
thresh2 = np.where(gray_array > T2, 255, 0).astype(np.uint8)
thresh3 = np.where(gray_array > T3, 255, 0).astype(np.uint8)

# Tampilkan hasil ketiga citra biner tersebut secara berdampingan menggunakan plt.subplots[cite: 9]
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Menampilkan biner T = 80
axes[0].imshow(thresh1, cmap='gray', vmin=0, vmax=255)
axes[0].set_title(f"Biner (T = {T1})")
axes[0].axis('off')

# Menampilkan biner T = 128
axes[1].imshow(thresh2, cmap='gray', vmin=0, vmax=255)
axes[1].set_title(f"Biner (T = {T2})")
axes[1].axis('off')

# Menampilkan biner T = 200
axes[2].imshow(thresh3, cmap='gray', vmin=0, vmax=255)
axes[2].set_title(f"Biner (T = {T3})")
axes[2].axis('off')

plt.tight_layout()
plt.show()
```

### Soal Praktikum 2: Menampilkan Channel Warna Tunggal (Color Channel Isolation)

**Instruksi:**
1. Buatlah program untuk mengisolasi kanal warna **Biru (Blue)** saja dari citra yang sudah diinputkan.
2. Ingat bahwa untuk menampilkan kanal biru murni (bukan grayscale), Anda harus menyetel nilai kanal Merah (R) dan Hijau (G) menjadi `0` pada array citra asli.
3. Tampilkan citra asli berdampingan dengan citra hasil isolasi kanal biru tersebut.

**CONTOH HASIL**
image.png

```python
# @title Code jawaban Tugas Praktikum 2

#tulis disini

import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Membaca gambar (misal: 'Matahari.jpg') sebagai RGB
img = np.array(Image.open('Matahari.jpg').convert('RGB'))

# Membuat salinan array untuk mengisolasi setiap kanal warna murni
# (Menyetel kanal lain menjadi 0 agar tampil sesuai warna aslinya, bukan grayscale)
img_R = img.copy()
img_R[:, :, 1] = 0  # Nolkan Hijau
img_R[:, :, 2] = 0  # Nolkan Biru

img_G = img.copy()
img_G[:, :, 0] = 0  # Nolkan Merah
img_G[:, :, 2] = 0  # Nolkan Biru

img_B = img.copy()
img_B[:, :, 0] = 0  # Nolkan Merah
img_B[:, :, 1] = 0  # Nolkan Hijau

# Membuat visualisasi 2 baris dan 2 kolom menggunakan plt.subplots
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# Citra RGB asli
axes[0, 0].imshow(img)
axes[0, 0].set_title("Citra Asli (RGB)")

# Kanal Merah murni
axes[0, 1].imshow(img_R)
axes[0, 1].set_title("Hanya Kanal Red (R)")

# Kanal Hijau murni
axes[1, 0].imshow(img_G)
axes[1, 0].set_title("Hanya Kanal Green (G)")

# Kanal Biru murni
axes[1, 1].imshow(img_B)
axes[1, 1].set_title("Hanya Kanal Blue (B)")

# Menyembunyikan sumbu koordinat untuk semua subplot
for ax in axes.ravel():
    ax.axis('off')

plt.tight_layout()
plt.show()
```
